# numpy 


#   1.Arry Creation

```python
arr=np.array([1,2,3,4,5,6])
arr2d=np.array([[1,2,3,4,5],[6,7,7,8,9]])
zero_arr= np.zeros((2,3))
ones_arr=np.ones((2,2))
identity=np.eye(3)
print(identity)
```

#   2.Array properties

```python
print(ones_arr.shape)
print(ones_arr.size)
```


#  3.sequence generation

```python

lin=np.linspace(0,100,20)
ran=np.arange(0,100,20)
print(lin)
print(ran)
```


#  4. slicing & indexing 

```python
print(ones_arr.dtype)
print(ones_arr.ndim)
print(arr[0])
print(arr[1:3])
print(arr2d[:,3:])
```


#  5.Boolean

```python
print(arr[arr < 4])
print(arr< 4)
print(arr[[0,1,2]])
```


#  6.vectorized operations

```python
print(arr+5)
print(arr**2)
print(np.sqrt(arr))
A= np.array([[1,2],[3,4]])
B= np.array([[5,6],[7,8]])
AB=A@B
print(AB)
```



#  7.Aggregation & statistics

```python
print(np.sum(arr))
print(np.mean(arr))
print(np.std(arr)) 
```

#  8.Reshapeing

```python
print(arr.reshape(2,3))
print(arr.T)
```

## 9.Stacking

```python
a=np.array([1,2,3])
b=np.array([4,5,6])
print(np.vstack([a,b]))
print(np.hstack([a,b]))
arr=np.arange(9)
print(np.split(arr,3))
```
